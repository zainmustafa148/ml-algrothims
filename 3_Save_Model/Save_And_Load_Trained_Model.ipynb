# Save and Load a Trained Model
This notebook trains a linear regression model on home prices, then saves it to a file with two methods (`pickle` and `joblib`) and loads it back to confirm it gives the same predictions.

**Note:** Upload `homeprices.csv` to Colab's Files section (`/content/`) before running.

**Security note:** `pickle.load` and `joblib.load` can run arbitrary code hidden inside a file. Only load model files that you created yourself or that come from a source you fully trust.

## 1. Import libraries

```python
# Handle data as tables (DataFrames)
import pandas as pd
# Numerical operations (not used directly in this notebook, but common in ML)
import numpy as np
# Linear regression model from scikit-learn
from sklearn import linear_model
```

## 2. Load the data

```python
# Read the CSV file into a DataFrame
# (the file must be in Colab's /content/ folder, otherwise FileNotFoundError is raised)
df = pd.read_csv("homeprices.csv")
# Show the first 5 rows to check the data
df.head()
```

## 3. Train the model

```python
# Create linear regression object
model = linear_model.LinearRegression()
# Train the model: X = area (2D input), y = price
model.fit(df[['area']],df.price)
```

## 4. Make a prediction

```python
# Predict the price of a home with area = 5000 sq ft
# Original line: model.predict([[5000]])
# The model was trained on a DataFrame with column name 'area', so passing the input
# with the same column name avoids sklearn's "feature names" warning.
# The logic and the result are exactly the same.
model.predict(pd.DataFrame([[5000]], columns=['area']))
```

Save Model To a File Using Python Pickle

```python
# pickle is Python's built-in module for saving objects to a file
import pickle
```

```python
# Save (serialize) the trained model to a file named 'model_pickle'
# 'wb' = write in binary mode; the 'with' block closes the file automatically
with open('model_pickle','wb') as file:
    pickle.dump(model,file)
```

```python
# Load (deserialize) the model back from the file
# 'rb' = read in binary mode
# Security: pickle.load can execute malicious code, so only load files you created or trust.
with open('model_pickle','rb') as file:
    mp = pickle.load(file)
```

```python
# Predict using the loaded model; it should match the original model's prediction
# Original line: mp.predict([[5000]])
# (DataFrame input with the same column name avoids the "feature names" warning)
mp.predict(pd.DataFrame([[5000]], columns=['area']))
```

Save Trained Model Using joblib

```python
# joblib is commonly used for scikit-learn models (efficient with large NumPy arrays)
import joblib
```

```python
# Save the trained model to a file named 'model_joblib'
joblib.dump(model, 'model_joblib')
```

```python
# Load the model back from the file
# Security: like pickle, joblib.load can execute malicious code, so only load trusted files.
mj = joblib.load('model_joblib')
```

```python
# Predict using the loaded model; it should match the original model's prediction
# Original line: mj.predict([[5000]])
# (DataFrame input with the same column name avoids the "feature names" warning)
mj.predict(pd.DataFrame([[5000]], columns=['area']))
```
