# Multiclass Classification with Logistic Regression: Iris Dataset
This notebook trains a logistic regression model to classify iris flowers into 3 species (*setosa*, *versicolor*, *virginica*) using 4 measurements: sepal length, sepal width, petal length and petal width.

**Note:** The Iris dataset is built into scikit-learn, so no CSV file needs to be uploaded.

## 1. Import libraries

```python
# Built-in Iris flower dataset
from sklearn.datasets import load_iris
# Logistic regression classifier (it handles multiple classes automatically)
from sklearn.linear_model import LogisticRegression
# Helper to split data into training and test sets
from sklearn.model_selection import train_test_split
```

## 2. Load and explore the data

```python
# Load the dataset (a Bunch object holding data, target and target_names)
iris = load_iris()
# Feature values: 150 flowers x 4 measurements (in cm)
iris.data
```

```python
# Target labels as numbers: 0, 1 or 2 (one per flower)
iris.target
```

```python
# Names of the 3 classes; the number in 'target' is the index into this list
iris.target_names
```

## 3. Create the model

```python
# Create the logistic regression model
# max_iter=1000 gives the solver enough steps to converge (the default 100 can show a ConvergenceWarning)
model = LogisticRegression(max_iter=1000)
```

## 4. Split into training and test sets

```python
# 80% training, 20% testing
# No random_state is set, so the split (and the score) can change on every run.
X_train, X_test, y_train, y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.2
)
```

## 5. Train and evaluate

```python
# Train the model using ONLY the training data
model.fit(X_train, y_train)
```

```python
# Accuracy on the unseen test data (1.0 = all predictions correct)
model.score(X_test, y_test)
```

## 6. Predictions

```python
# Predict the class numbers for the first 5 test flowers
model.predict(X_test[0:5])
```

```python
#Actual Values
# The real class numbers for the same 5 flowers, to compare with the predictions above
y_test[0:5]
```

```python
# New Prediction
# Predict the class numbers for the first 10 test flowers
predictions = model.predict(X_test[0:10])

print(predictions)
```

```python
# Convert the predicted class numbers into species names
print(iris.target_names[predictions])
```

```python
# Side-by-side comparison of predicted and actual species names (first 10 test flowers)
print("Predicted:", iris.target_names[predictions], "\n\n")
print("Actual:   ", iris.target_names[y_test[0:10]])
```
