# Multivariate Linear Regression: Hiring Salary
This notebook predicts *salary ($)* from three features: *experience*, *test score (out of 10)* and *interview score*, using multivariate linear regression (`y = m1*x1 + m2*x2 + m3*x3 + b`).

**Note:** Upload `hiring.csv` to Colab's Files section (`/content/`) before running.

## 1. Import libraries

```python
# Handle data as tables (DataFrames)
import pandas as pd
# Numerical operations (not used directly in this notebook, but common in ML)
import numpy as np
# Linear regression model from scikit-learn
from sklearn import linear_model
```

## 2. Install and import word2number
Some values in the `experience` column are written as words (e.g. "five"). `word2number` converts them into numbers.

```python
# Install the third-party package from PyPI (needed only once per Colab session)
# Security tip: for reproducible and safer installs, pin the version, e.g. !pip install word2number==1.1
!pip install word2number
```

```python
# Converts number words like "five" into the number 5
from word2number import w2n
```

## 3. Load the data

```python
# Read the CSV file into a DataFrame
# (the file must be in Colab's /content/ folder, otherwise FileNotFoundError is raised)
df = pd.read_csv('hiring.csv')
# Colab automatically displays the last expression in a cell
df
```

Data Preprocessing

```python
# Convert word values in 'experience' (e.g. "five") to numbers.
# If the value is already a number or NaN (not a string), keep it as it is.
df['experience'] = df['experience'].apply(
    lambda x: w2n.word_to_num(x) if isinstance(x, str) else x
)
```

```python
# Check the median of 'experience' (used below to fill missing values)
df.experience.median()
```

```python
# Fill missing 'experience' values with the median
median_experience = df['experience'].median()
df['experience'] = df['experience'].fillna(median_experience)
```

```python
# Fill missing test scores with the median test score
median_test_score = df['test_score(out of 10)'].median()

df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(
    median_test_score
)
```

```python
# Check the cleaned data (no missing values should remain)
df
```

## 4. Train the model

```python
# Create linear regression object
reg = linear_model.LinearRegression()

# Train the model:
# X (features) = all columns except 'salary($)'
# y (target)   = 'salary($)'
reg.fit(df.drop('salary($)',axis='columns'),df['salary($)'])
```

## 5. Make predictions

2 yr experience, 9 test score, 6 interview score

```python
# Original line: reg.predict([[2, 9, 6]])
# The model was trained on a DataFrame with named columns, so passing the input
# with the same column names avoids sklearn's "feature names" warning.
# The logic and the result are exactly the same.
reg.predict(pd.DataFrame([[2, 9, 6]], columns=df.drop('salary($)',axis='columns').columns))
```

12 yr experience, 10 test score, 10 interview score

```python
# Original line: reg.predict([[12, 10, 10]])
# Same fix as above: use the same column names as the training data.
reg.predict(pd.DataFrame([[12, 10, 10]], columns=df.drop('salary($)',axis='columns').columns))
```
