# Linear Regression: Canada Per Capita Income
This notebook predicts *per capita income (US$)* from *year* using simple linear regression (`y = m*x + b`).

**Note:** Upload `canada_per_capita_income (1).csv` to Colab's Files section (`/content/`) before running.

## 1. Import libraries

```python
# Handle data as tables (DataFrames)
import pandas as pd
# Numerical operations (not used directly in this notebook, but common in ML)
import numpy as np
# Linear regression model from scikit-learn
from sklearn import linear_model
# Plotting and graphs
import matplotlib.pyplot as plt
```

## 2. Load the data

```python
# Read the CSV file into a DataFrame
# (the file must be in Colab's /content/ folder, otherwise FileNotFoundError is raised)
df = pd.read_csv('canada_per_capita_income (1).csv')
# Colab automatically displays the last expression in a cell
df
```

## 3. Visualize the data

```python
# Show plots inline inside the notebook
%matplotlib inline
plt.xlabel('year')                          # X-axis label
plt.ylabel('per capita income (US$)')       # Y-axis label
# Scatter plot: one red '+' marker per year showing income
plt.scatter(df.year, df['per capita income (US$)'] ,color='red',marker='+')
```

## 4. Separate features (X) and target (y)

```python
# X (input feature): drop the income column so only 'year' remains
new_df = df.drop('per capita income (US$)',axis='columns')
new_df
```

```python
# y (target/label): the value we want to predict, i.e. per capita income
# (the variable is named 'price' but it holds income here; the original name is kept)
price = df['per capita income (US$)']
price
```

## 5. Train the model

```python
# Create linear regression object
reg = linear_model.LinearRegression()
# Train the model: learn the relationship between year (new_df) and income (price)
reg.fit(new_df,price)
```

## 6. Make a prediction

```python
# Predict per capita income for the year 2020
# Original line: reg.predict([[2020]])
# The model was trained on a DataFrame with column name 'year', so passing the input
# with the same column name avoids sklearn's "feature names" warning.
# The logic and the result are exactly the same.
reg.predict(pd.DataFrame([[2020]], columns=new_df.columns))
```

## 7. Model parameters (m and b)

```python
# Coefficient (m): the slope, i.e. how much income rises on average per year
reg.coef_
```

```python
# Intercept (b): the value of the line when year = 0
reg.intercept_
```
