# Train Test Split: Car Prices
This notebook shows how to split data into a training set and a test set with `train_test_split`, train a linear regression model on the training part, and predict on the unseen test part.

**Note:** Upload `carprices (1).csv` to Colab's Files section (`/content/`) before running.

## 1. Import libraries and load the data

```python
# Handle data as tables (DataFrames)
import pandas as pd
```

```python
# Read the CSV file into a DataFrame
# (the file must be in Colab's /content/ folder, otherwise FileNotFoundError is raised)
df = pd.read_csv("carprices (1).csv")
# Show the first 5 rows to check the data
df.head()
```

## 2. Separate features (X) and target (y)

```python
# X (input features): the columns used to make the prediction
X = df[['Mileage','Age(yrs)']]
```

```python
# y (target/label): the value we want to predict, the selling price
y = df['Sell Price($)']
```

## 3. Split into training and test sets

```python
# Split the data: 80% for training, 20% for testing (test_size=0.2)
# The split is random, so rows (and results) change on every run
# because no random_state is set.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
# Features used to train the model
X_train
```

```python
# Features kept aside to test the model (not used in training)
X_test
```

```python
# Prices (labels) that go with X_train
y_train
```

```python
# Prices (labels) that go with X_test (the real answers to compare predictions against)
y_test
```

## 4. Train the model and predict

```python
# Linear regression model from scikit-learn
from sklearn.linear_model import LinearRegression
# Create the model object
clf = LinearRegression()
# Train the model using ONLY the training data
clf.fit(X_train, y_train)
```

```python
# Predict prices for the test set; compare with y_test to see how well the model does
clf.predict(X_test)
```

## 5. Reproducible split
Setting `random_state` makes the split the same every time you run the code.

```python
# New split: 70% training, 30% testing
# random_state=10 fixes the shuffle, so you always get the same rows in each set.
# Note: this overwrites X_train, X_test, y_train and y_test from the earlier split,
# but the model 'clf' above was already trained on the old split and is not retrained here.
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=10)
X_test
```
