# K-Fold Cross Validation: Comparing Models on the Iris Dataset
This notebook uses `cross_val_score` (K-Fold cross validation) to compare 4 classifiers on the Iris dataset: Logistic Regression, Decision Tree, SVM and Random Forest.

Instead of one train/test split, the data is divided into 5 folds (the default). The model is trained on 4 folds and tested on the remaining 1, and this is repeated 5 times so every fold is used for testing once. The average of the 5 scores gives a more reliable accuracy estimate.

**Note:** The Iris dataset is built into scikit-learn, so no CSV file needs to be uploaded.

## 1. Import libraries

```python
# Numerical operations (used below for np.average)
import numpy as np
# Built-in Iris flower dataset
from sklearn.datasets import load_iris
# Runs K-Fold cross validation and returns one accuracy score per fold
from sklearn.model_selection import cross_val_score
# The 4 classifiers to compare
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
```

## 2. Load the data

```python
# Load the dataset (150 flowers, 4 measurements each, 3 species)
iris = load_iris()
```

Logistic regression

```python
# Cross validation with Logistic Regression: returns 5 accuracy scores (one per fold)
# Note: LogisticRegression() uses max_iter=100 by default, which can show a ConvergenceWarning on this data.
# The scores are still valid; max_iter=1000 would remove the warning (suggestion only, code not changed).
lg_scores = cross_val_score(LogisticRegression(), iris.data, iris.target)
lg_scores
```

```python
# Average accuracy across the 5 folds
np.average(lg_scores)
```

Decision Tree

```python
# Cross validation with a Decision Tree: 5 accuracy scores (one per fold)
dt_scores = cross_val_score(DecisionTreeClassifier(), iris.data, iris.target)
dt_scores
```

```python
# Average accuracy across the 5 folds
np.average(dt_scores)
```

svm

```python
# Cross validation with an SVM (SVC, default RBF kernel): 5 accuracy scores (one per fold)
svm_scores = cross_val_score(SVC(), iris.data, iris.target)
svm_scores
```

```python
# Average accuracy across the 5 folds
np.average(svm_scores)
```

Random Forest

```python
# Cross validation with a Random Forest of 40 trees: 5 accuracy scores (one per fold)
# The forest is random, so scores can change slightly on every run (no random_state is set).
rf_scores = cross_val_score(RandomForestClassifier(n_estimators=40), iris.data, iris.target)
rf_scores
```

```python
# Average accuracy across the 5 folds
np.average(rf_scores)
```
