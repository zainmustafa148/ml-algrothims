# Support Vector Machine (SVM): Handwritten Digits
This notebook classifies images of handwritten digits (0 to 9) using an SVM with the RBF kernel.

**Note:** The digits dataset is built into scikit-learn, so no CSV file needs to be uploaded.

## 1. Import libraries

```python
# Handle data as tables (DataFrames)
import pandas as pd
```

## 2. Load and explore the data

```python
# Built-in dataset of 8x8 pixel images of handwritten digits
# Each image is flattened into 64 pixel values (1797 images in total)
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
# The actual digit (0 to 9) shown in each image
digits.target
```

```python
# Names of the classes: the digits 0 to 9
digits.target_names
```

```python
# Put the 64 pixel values of each image into a DataFrame
# Note: the second argument sets the row INDEX to the digit labels (it is not the column names).
# The index values repeat, but this does not affect the model because the labels are added as a 'target' column below.
df = pd.DataFrame(digits.data,digits.target)
df.head()
```

```python
# Add the digit labels as a separate column named 'target'
df['target'] = digits.target
df.head(20)
```

## 3. Split into training and test sets

```python
# X = the 64 pixel columns (everything except 'target'), y = the 'target' column (the digit)
# 80% training, 20% testing
# No random_state is set, so the split (and the accuracy) can change on every run.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('target',axis='columns'), df.target, test_size=0.2)
```

## 4. Create the SVM model

```python
# SVC = Support Vector Classifier
# kernel='rbf' (Radial Basis Function) lets the model draw curved (non-linear) boundaries between classes
from sklearn.svm import SVC
rbf_model = SVC(kernel='rbf')
```

## 5. Train and evaluate

```python
# Number of images used for training
len(X_train)
```

```python
# Number of images used for testing
len(X_test)
```

```python
# Train the model using ONLY the training data
rbf_model.fit(X_train, y_train)
```

```python
# Accuracy on the unseen test data (1.0 = all predictions correct)
rbf_model.score(X_test,y_test)
```
